```python
import cv2
```

```python
from google.colab.patches import cv2_imshow

vid = cv2.VideoCapture(r"video.mp4")
while vid.isOpened():
  ret, frame = vid.read()
  if not ret:
    print(f"Video not available or ended.")
    break

  cv2_imshow(frame)
  if cv2.waitKey(1) & 0xFF == ord("q"):
    break

vid.release()
cv2.destroyAllWindows()
```
